```python
def convert_emp_length(x):
    if pd.isnull(x):
        return None
    if x == "< 1 year":
        return 0
    if x == "10+ years":
        return 10
    return int(x.split()[0])
```

```python
bank_ds["emp_length_num"] = bank_ds["emp_length"].apply(convert_emp_length)
```

```python
bank_ds.isnull().sum()
```

```python
bank_ds["emp_length_num"].fillna(-1, inplace=True)  ##I encoded missing values as -1 to represent an unknown category, preserving information without distorting the ordinal structure
```

```python
bank_ds["emp_length_num"].isnull().sum()
```

```python
bank_ds.drop("emp_length", axis=1, inplace=True)
```

```python
bank_ds["emp_length_num"].value_counts()
```
